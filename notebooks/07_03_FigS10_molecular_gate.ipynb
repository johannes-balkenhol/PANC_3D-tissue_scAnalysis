```python
import os
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, Circle, FancyArrowPatch

# Output directory (relative to project root; adjust if you run elsewhere)
OUTDIR = 'figures_submission/supplementary/FigS11_molecular_gate_schematic'
os.makedirs(OUTDIR, exist_ok=True)

plt.rcParams.update({'font.family': 'DejaVu Sans', 'svg.fonttype': 'none'})
```

```python
# ---- palette -------------------------------------------------------------
GREEN_F, GREEN_E, GREEN_T = '#E3F1E7', '#3f9d63', '#1f6b3b'   # activating / S-phase
RED_F,   RED_E,   RED_T   = '#FBE9E7', '#c0392b', '#a93226'   # inhibitory
CDK1_F,  CDK1_E,  CDK1_T  = '#E7ECF5', '#2c3e50', '#1b2631'   # convergence node
OUT_F,   OUT_E,   OUT_T   = '#FCEFD2', '#e0a12f', '#7a5a12'   # output state
ACT, INH, AMBER = '#2e8b57', '#c0392b', '#c0740f'


def node(ax, x, y, w, h, label, sub=None, fc=GREEN_F, ec=GREEN_E, tc=GREEN_T, r=0.14):
    ax.add_patch(FancyBboxPatch((x-w/2, y-h/2), w, h,
                 boxstyle=f'round,pad=0.01,rounding_size={r}',
                 fc=fc, ec=ec, lw=1.8, zorder=3))
    dy = 0.06 if sub else 0
    ax.text(x, y+dy, label, ha='center', va='center', fontsize=12.5,
            fontweight='bold', color=tc, zorder=4)
    if sub:
        ax.text(x, y-0.17, sub, ha='center', va='center', fontsize=8.5, color=tc, zorder=4)


def circle(ax, x, y, rad, label, sub):
    ax.add_patch(Circle((x, y), rad, fc=CDK1_F, ec=CDK1_E, lw=2.4, zorder=3))
    ax.text(x, y+0.10, label, ha='center', va='center', fontsize=17,
            fontweight='bold', color=CDK1_T, zorder=4)
    ax.text(x, y-0.22, sub, ha='center', va='center', fontsize=8.5, color='#5d6d7e', zorder=4)


def activate(ax, p0, p1, rad=0.0, color=ACT, lw=2.2):
    ax.add_patch(FancyArrowPatch(p0, p1, connectionstyle=f'arc3,rad={rad}',
                 arrowstyle='-|>', mutation_scale=18, lw=lw, color=color,
                 shrinkA=6, shrinkB=8, zorder=2))


def inhibit(ax, p0, p1, rad=0.0, color=INH, lw=2.2, bar=0.16):
    ax.add_patch(FancyArrowPatch(p0, p1, connectionstyle=f'arc3,rad={rad}',
                 arrowstyle='-', lw=lw, color=color, shrinkA=6, shrinkB=10, zorder=2))
    x1, y1 = p1
    ax.plot([x1, x1], [y1-bar, y1+bar], color=color, lw=3.2, zorder=2, solid_capstyle='round')
```

```python
fig, ax = plt.subplots(figsize=(10, 7.5))
ax.set_xlim(0, 10); ax.set_ylim(0, 7.5); ax.axis('off')

# ---- titles --------------------------------------------------------------
ax.text(5.0, 7.28, 'CDK1\u2013CDKN1A\u2013WEE1 Molecular Gate', ha='center',
        fontsize=16, fontweight='bold', color='#1b2631')
ax.text(5.0, 7.00, 'Two orthogonal inputs converge on a single rate-limiting checkpoint',
        ha='center', fontsize=10, color='#34495e')
ax.text(5.0, 6.78, 'Convergence is emergent: detected specifically in the combined '
        'G1\u2192S + TGF-\u03b21 state,\nnot present when either program acts alone.',
        ha='center', fontsize=8.5, style='italic', color=AMBER)

# ---- column headers ------------------------------------------------------
ax.text(2.3, 6.35, 'S-phase progression signals', ha='center', fontsize=11,
        fontweight='bold', color='#1b2631')
ax.text(2.3, 6.14, '(independent contributors, not a validated sequence)',
        ha='center', fontsize=8, color='#5d6d7e')
ax.text(7.7, 6.35, 'TGF-\u03b21 signaling', ha='center', fontsize=11,
        fontweight='bold', color='#1b2631')

# ---- nodes ---------------------------------------------------------------
node(ax, 2.3, 5.70, 1.7, 0.6, 'E2F1')
node(ax, 2.3, 4.85, 1.7, 0.6, 'MYBL2')
node(ax, 2.3, 4.00, 1.7, 0.6, 'CCNA2')
node(ax, 7.7, 5.70, 1.7, 0.6, 'TGF-\u03b21')
node(ax, 7.7, 4.85, 1.7, 0.6, 'SMAD3')
node(ax, 7.7, 3.85, 1.7, 0.62, 'CDKN1A', '(p21)', fc=RED_F, ec=RED_E, tc=RED_T)
node(ax, 1.55, 2.55, 1.6, 0.62, 'WEE1', 'inhibitory kinase', fc=RED_F, ec=RED_E, tc=RED_T)
circle(ax, 5.0, 2.95, 0.78, 'CDK1', 'convergence point')
ax.add_patch(FancyBboxPatch((3.0, 0.55), 4.0, 0.85, boxstyle='round,pad=0.01,rounding_size=0.12',
             fc=OUT_F, ec=OUT_E, lw=1.8, zorder=3))
ax.text(5.0, 1.10, 'S-phase persistence state', ha='center', fontsize=11.5,
        fontweight='bold', color=OUT_T, zorder=4)
ax.text(5.0, 0.83, 'chemo-tolerant subpopulation', ha='center', fontsize=9, color=OUT_T, zorder=4)

# ---- edges ---------------------------------------------------------------
activate(ax, (2.3, 5.40), (2.3, 5.15))                 # E2F1 -> MYBL2
activate(ax, (3.05, 4.72), (4.35, 3.30), rad=-0.15)    # MYBL2 -> CDK1
activate(ax, (3.05, 3.95), (4.30, 2.75), rad=0.18)     # CCNA2 -> CDK1
activate(ax, (7.7, 5.40), (7.7, 5.15))                 # TGF-b1 -> SMAD3
activate(ax, (7.7, 4.55), (7.7, 4.16))                 # SMAD3 -> CDKN1A
inhibit(ax, (6.95, 3.70), (5.70, 3.15), rad=0.18)      # CDKN1A -| CDK1
inhibit(ax, (2.30, 2.55), (4.24, 2.80), rad=-0.15)     # WEE1 -| CDK1
activate(ax, (5.0, 2.17), (5.0, 1.45), lw=2.4, color='#1b2631')  # CDK1 -> output

# ---- edge annotations ----------------------------------------------------
ax.text(3.9, 3.55, 'activates \u2191', fontsize=8.5, style='italic', color=ACT, ha='center')
ax.text(3.35, 3.10, 'activates \u2191', fontsize=8.5, style='italic', color=ACT, ha='center')
ax.text(6.55, 3.62, 'inhibits \u22a3', fontsize=8.5, style='italic', color=INH, ha='center')
ax.text(3.15, 2.90, 'maintains arrest', fontsize=8.5, style='italic', color=INH, ha='center')
ax.text(8.55, 4.55, '(acts with SMAD4;\nco-factor omitted for clarity)', fontsize=7.5,
        style='italic', color='#5d6d7e', ha='left')
ax.text(8.55, 3.45, 'induction is p53-independent', fontsize=7.5, style='italic',
        color=RED_T, ha='left')

# ---- legend --------------------------------------------------------------
ax.annotate('', xy=(0.85, 0.32), xytext=(0.35, 0.32),
            arrowprops=dict(arrowstyle='-|>', color=ACT, lw=2.4, mutation_scale=16))
ax.text(0.95, 0.32, 'Activation', va='center', fontsize=9, color='#1b2631')
ax.plot([2.55, 2.55], [0.20, 0.44], color=INH, lw=3.2, solid_capstyle='round')
ax.plot([2.10, 2.55], [0.32, 0.32], color=INH, lw=2.4)
ax.text(2.70, 0.32, 'Inhibition', va='center', fontsize=9, color='#1b2631')

plt.tight_layout()
```
